##**3. Menampilkan Citra KTM**

```python
FILE_KTM = "KTM.jpeg"
img = cv.imread(FILE_KTM)
cv2_imshow(img)

imgRGB = cv.cvtColor(img, cv.COLOR_BGR2RGB)  # Konversi BGR ke RGB
plt.imshow(imgRGB)
plt.show()
```

##**4. Menutup Bagian Tertentu pada KTM**

```python
img_tertutup = img.copy()

# AREA 1: Tutup tulisan "POLITEKNIK NEGERI MALANG" (RED)
pt1_1, pt2_1 = (180, 50), (600, 160)
warna_1 = (0, 0, 255) # Merah
cv.rectangle(img_tertutup, pt1=pt1_1, pt2=pt2_1, color=warna_1, thickness=-1)

# AREA 2 : Bagian Bank BNI (GREEN)
pt1_2, pt2_2 = (1050, 50), (1450, 160)
warna_2 = (0, 255, 0) # Hijau
cv.rectangle(img_tertutup, pt1=pt1_2, pt2=pt2_2, color=warna_2, thickness=-1)

# AREA 3 : Bagian Prodi (BLUE)
pt1_3, pt2_3 = (350, 900), (850, 970)
warna_3 = (255, 0, 0) # Biru
cv.rectangle(img_tertutup, pt1=pt1_3, pt2=pt2_3, color=warna_3, thickness=-1)

# Tampilkan hasil
imgRGB_tertutup = cv.cvtColor(img_tertutup, cv.COLOR_BGR2RGB)
plt.imshow(imgRGB_tertutup)
plt.title("KTM setelah bagian tertentu ditutup")
plt.show()
```

```python
img_coba = imgRGB_tertutup.copy()

plt.imshow(img_coba)
plt.xticks(range(0, img_coba.shape[1], 50))
plt.yticks(range(0, img_coba.shape[0], 50))
plt.show()
```
